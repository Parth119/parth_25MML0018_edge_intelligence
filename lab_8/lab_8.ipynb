```python
import os
os.environ['KMP_DUPLICATE_LIB_OK'] = 'True'

import cv2
from ultralytics import YOLO

# 1. Load the YOLOv8 model (pretrained)
model = YOLO('yolov8n.pt') 

# 2. Open the video file
video_path = "output.mp4"
cap = cv2.VideoCapture(video_path)

# Loop through the video frames
while cap.isOpened():
    success, frame = cap.read()
    if success:
        # 3. Run YOLOv8 tracking on the frame
        # 'persist=True' tells the tracker to remember objects from the last frame
        results = model.track(frame, persist=True, tracker="botsort.yaml")

        # 4. Visualize the results on the frame
        annotated_frame = results[0].plot()

        # 5. Display the annotated frame
        cv2.imshow("YOLOv8 Tracking", annotated_frame)

        # Break the loop if 'q' is pressed
        if cv2.waitKey(1) & 0xFF == ord("q"):
            break
    else:
        break

cap.release()
cv2.destroyAllWindows()
```
